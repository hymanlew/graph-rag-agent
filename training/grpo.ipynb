## 功能说明

此脚本用于使用GRPO（Graph RAG Pre-training Optimization）方法训练Qwen2-3B模型，
使其更好地适应知识图谱中的实体抽取任务。训练后的模型将能够：

- 更准确地从文本中识别和抽取实体
- 理解实体之间的关系类型
- 适应本项目特定的领域知识（如华东理工大学学生手册相关内容）
- 提高知识图谱构建的质量和效率

## 训练目标

训练的核心目标是优化模型的实体抽取能力，使其产出的实体和关系更符合知识图谱的要求，
为整个图数据库的构建提供更可靠的基础数据。

## 与项目的关系

训练后的模型将被用于：

- `build/build_graph.py` 中的实体抽取流程
- `graph/extraction` 模块的关系识别
- 支持增量更新知识图谱的实体和关系
